```python
df = spark.read.csv(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep = '|', quote= "'"
)
```

```python
df.na.drop(how = "any").display()
```

```python
df.na.drop(how ="all").display()
```

```python
df.na.drop(subset = ["id","name"]).display()
```

```python
df.na.drop(subset = ["id","name"],how = "all").display()

```

```python
df.na.fill("NULL IN SOURCE").na.fill(
    -1,
    subset=["id"],
).na.fill(-0,subset=["exp"]).display()
```

```python
from pyspark.sql.functions import avg
avg_exp = df.select(avg("exp")).collect()[0][0]
```

```python
dic = {
    "name":"Annanymous",
    "id":-1,
    "exp":avg_exp,

}

df.na.fill(dic).display()
```

```python
df.withColumn(
    "doj", when(col("doj").isNull(), current_date()).otherwise(col("doj"))
).display()
```
