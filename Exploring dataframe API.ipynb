```python
df = spark.read.csv("/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/employee.csv",header = True,inferSchema = True,sep = '|',quote= "'")
df.display()
```

```python
from pyspark.sql.functions import col
df.select(col("id"),col("name").alias("full name")).display()
```

```python
df.filter(col("gen")=='M').filter(col("company")=="Infosys").select(col("name")).display()

df.filter(col("gen" == "M") & (col("company") == "Infosys")).select(col("name")).display()

```

```python
df.filter((col("company") == "Cisco") | (col("company") == "cisco")).select(col("name")).display()
```

```python
from pyspark.sql.functions import lower
df.filter(lower("company") =='cisco').select("name").display()
```

```python
from pyspark.sql.functions import lit
df.withColumn("is_employeed",lit("True")).display()
```

```python
from pyspark.sql.functions import when

df.withColumn(
    "new_exp_category",
    when(col("exp") >= 10) & (col("gen") == "M","senior")
    .when(col("exp") >= 9) & col("gen") == "F","senior")
    .when(col("exp") >= 5) & col("gen") == "M","mid")
    .when(col("exp") >= 4)) & col("gen") == "F","mid")
    .when(col("exp") >= 0) & col("gen") == "M","junior")
    .when(col("exp") >= 0) & col("gen") == "M","junior")
    .otherwise("invalid experience"),
).display()
```
