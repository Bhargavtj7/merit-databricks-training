## Prerequisites
Load "users_001.csv" into Data Frame

```python
df=spark.read.csv(path="/Volumes/merit_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_002.csv",header=True,inferSchema=True)
df.limit(4).display()
```

reading a csv file and writing it into parquet format (which isn't the delta format)

```python
df.write.format("parquet").save(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet",
    mode="overwrite",
)
df_parquet = spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
)
df_parquet.limit(4).display()
```

## create a view

```python
df_parquet.createOrReplaceTempView("user_vw")
```

# Drawback - Updates are not allowed

```python
%sql
UPDATE user_vw
SET country="Bharat"
WHERE country="India"
```

this is one drawback of cloud storage where update operation can't be performed

# Drawback - Overwrite isn't allowed because we can't rollback

```python
from pyspark.sql.functions import col
df.filter(col('country')=='India').write.format("parquet").save(
    path="/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet",
    mode="overwrite",
)
```

```python
spark.read.parquet(
    "/Volumes/merit_catalog/quickstart_schema/sandbox/output/output_parquet"
).limit(6).display()
```

now the data records other than country as india are gone and no roll back option
